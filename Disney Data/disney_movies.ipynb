```python
awards_df = pd.read_csv('data/disney_plus_shows.csv')
```

```python
awards_df = awards_df[awards_df['type'] == 'movie'].copy()
```

```python
awards_df.head()
```

```python
awards_df['special_wins'] = awards_df['awards'].str.extract(r'Won (\d+)', expand=False)
awards_df['special_nominations'] = awards_df['awards'].str.extract(r'Nominated for (\d+)',expand=False)
```

```python
awards_df['wins'] = awards_df['awards'].str.extract(r'(\d+) wins?',expand=False)
awards_df['nominations'] = awards_df['awards'].str.extract(r'(\d+) nominations?',expand=False)
```

```python
awards_df['wins'] = awards_df['wins'].fillna(0).astype(int)
awards_df['nominations'] = awards_df['nominations'].fillna(0).astype(int)
awards_df['special_wins'] = awards_df['special_wins'].fillna(0).astype(int)
awards_df['special_nominations'] = awards_df['special_nominations'].fillna(0).astype(int)
```

```python
awards_df[['special_wins','wins','special_nominations','nominations']]
```

```python
awards_df.info()
```

```python
awards_per_movie_df=awards_df[
    ['imdb_id','title','wins','special_wins','nominations','special_nominations']
    ]
```

```python
awards_per_movie_df.info()
```

```python
awards_per_movie_df.head()
```

```python
awards_per_movie_df['total_wins'] = awards_per_movie_df['wins'] + awards_per_movie_df['special_wins']
awards_per_movie_df['total_nominations'] = awards_per_movie_df['nominations'] + awards_per_movie_df['special_nominations']
```

```python
awards_per_movie_df = awards_per_movie_df.sort_values(['total_wins','total_nominations'], ascending = [False, False])
```

```python
awards_per_movie_df = awards_per_movie_df[
    [
        'imdb_id',
        'title',
        'total_wins',
        'total_nominations',
        'wins',
        'special_wins',
        'nominations',
        'special_nominations'
    ]
]
```

```python
awards_per_movie_df.head(30)
```

```python
awards_per_movie_df['imdb_rating']=awards_df['imdb_rating']
```

```python
awards_per_movie_df['total_wins'].corr(awards_per_movie_df['imdb_rating'])
```

```python
awards_per_movie_df['metascore']=awards_df['metascore']
```

```python
awards_per_movie_df['total_wins'].corr(awards_per_movie_df['metascore'])
```
